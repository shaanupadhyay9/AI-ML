## What is Regular Expression?

Regular Expression, commonly called Regex, is a sequence of characters used to 
search, match, extract, replace, and validate patterns in text.

Regex is particularly useful when working with unstructured text.

Common applications include:

- Extracting email addresses
- Extracting phone numbers
- Extracting dates
- Finding URLs
- Cleaning unwanted characters
- Validating usernames
- Validating passwords
- Finding numbers
- Replacing text patterns

```python
import re

text = "My email is shaan@example.com"

result = re.findall(r'\w+@\w+\.\w+', text)

print(result)
```

```python
text = "I have 25 apples and 10 oranges."

numbers = re.findall(r'\d+', text)

print(numbers)
```

```python
text = """
Contact us at shaan@gmail.com or support@company.org.
You can also email admin@test.co.in.
"""

emails = re.findall(
    r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}',
    text
)

print(emails)
```

```python
text = """
Contact numbers:
9876543210
9123456789
"""

phones = re.findall(r'\b[6-9]\d{9}\b', text)

print(phones)
```

```python
text = """
Call +91-9876543210
or +91 9123456789
"""

phones = re.findall(
    r'(?:\+91[- ]?)?[6-9]\d{9}',
    text
)

print(phones)
```

```python
text = """
Important dates:
10/08/2026
15-09-2026
01/10/2026
"""

dates = re.findall(
    r'\b\d{2}[-/]\d{2}[-/]\d{4}\b',
    text
)

print(dates)
```

```python
text = """
Visit https://www.google.com
or https://github.com
"""

urls = re.findall(
    r'https?://[^\s]+',
    text
)

print(urls)
```

```python
text = "Shaan works with Python and Pandas."

capital_words = re.findall(
    r'\b[A-Z][a-z]*\b',
    text
)

print(capital_words)
```

```python
text = "Hello!!! Welcome @#$ to Python."

clean_text = re.sub(
    r'[^A-Za-z0-9\s]',
    '',
    text
)

print(clean_text)
```

# Seaborn Visualization

Seaborn is a Python visualization library based on Matplotlib.

It is particularly useful for statistical data visualization.

Common Seaborn plots include:

- Bar plot
- Count plot
- Histogram
- Box plot
- Violin plot
- Scatter plot
- Line plot
- Heatmap
- Pair plot

```python
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Titanic-Dataset.csv")

df.head()
```

```python
sns.set_theme(style="whitegrid")
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='Sex'
)

plt.title("Passenger Count by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x='Pclass',
    y='Fare'
)

plt.title("Average Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x='Age',
    bins=30,
    kde=True
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x='Pclass',
    y='Age'
)

plt.title("Age Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x='Sex',
    y='Age'
)

plt.title("Age Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Age")
plt.show()
```

```python
plt.figure(figsize=(10, 7))

numeric_df = df.select_dtypes(include=np.number)

correlation = numeric_df.corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
pair_data = df[
    ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
].dropna()

sns.pairplot(
    pair_data,
    hue='Survived'
)

plt.show()
```

```python
age_survival = df.groupby('Age', as_index=False)['Survived'].mean()

plt.figure(figsize=(10, 5))

sns.lineplot(
    data=age_survival,
    x='Age',
    y='Survived'
)

plt.title("Survival Rate by Age")
plt.xlabel("Age")
plt.ylabel("Survival Rate")
plt.show()
```

```python
sns.catplot(
    data=df,
    x='Pclass',
    y='Fare',
    col='Sex',
    kind='box',
    height=5
)

plt.show()
```
